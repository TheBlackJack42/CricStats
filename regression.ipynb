```python
import numpy as np
from sklearn.linear_model import LinearRegression
import pandas as pd

from sklearn.neural_network import MLPRegressor

df = pd.read_csv('batsman_runs.csv')

df.sample(10)
```

```python
# REGRESSION
x = list(zip(df.balls, df.innings_runs, df.batting_position, df.runs_come_in, df.overs_come_in))
y = df.runs.to_list()

regr = MLPRegressor(max_iter=1000)
regr.fit(x, y)

preds, actuals = [], []

for xi, yi in zip(x, y):
    preds += [max(round(regr.predict([xi])[0],1), 0)]
    actuals += [yi]

df['predicted'] = preds
df['difference'] = df.runs - df.predicted
```

```python
# PLOT THE PREDICTED VS ACTUAL RUNS
#%matplotlib qt # uncomment to make the plot interactive in a separate window
import matplotlib.pyplot as plt
from seaborn import heatmap
from matplotlib.colors import LogNorm

freq_table = pd.crosstab(actuals, preds, colnames=['Actual'], rownames=['Predicted'])
frequency_table = freq_table.reindex(index=range(round(max(actuals))+1), columns=range(round(max(preds))+1), fill_value=0)

plt.figure(figsize=(10,10))

heatmap(frequency_table.iloc[::-1], norm=LogNorm())

plt.plot([171, 0], [0, 171], color='red')
```
